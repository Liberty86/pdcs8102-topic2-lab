Task 1.3

```python
def get_grade(score):
    if score>=80:
        return "Distinction"
    elif score>=70:
        return"Merit"
    elif score>=60:
        return"Pass"
    else:
        return "fail"
```

```python
print(get_grade(85))
```

```python
print(get_grade(72))
```

```python
print(get_grade(45))
```

```python
def calculate_average(scores):
    return sum(scores) / len(scores)
```

```python
scores = [75, 82, 68, 90,45]
```

```python
average=calculate_average(scores)
```

```python
print("average:", average)
```

```python
print("Grade:", get_grade(average))
```
